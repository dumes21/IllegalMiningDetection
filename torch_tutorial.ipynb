```python
import ee
import geemap

ee.Authenticate()
ee.Initialize()

region = ee.Geometry.Rectangle([
    -122.5, 37.7,
    -122.4, 37.8
])

l7 = (
    ee.ImageCollection("LANDSAT/LE07/C02/T1_L2")
    .filterBounds(region)
    .filterDate("2010-01-01", "2010-12-31")
    .sort("CLOUD_COVER")
    .first()
)

l7 = l7.select([
    "SR_B1",
    "SR_B2",
    "SR_B3",
    "SR_B4",
    "SR_B5",
    "SR_B7"
])

l8 = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
    .filterBounds(region)
    .filterDate("2015-01-01", "2015-12-31")
    .sort("CLOUD_COVER")
    .first()
)

l8 = l8.select([
    "SR_B2",
    "SR_B3",
    "SR_B4",
    "SR_B5",
    "SR_B6",
    "SR_B7"
])
```

```python
geemap.ee_export_image(
    l7,
    filename="landsat7.tif",
    scale=30,
    region=region
)

geemap.ee_export_image(
    l8,
    filename="landsat8.tif",
    scale=30,
    region=region
)
```
